```python
#kmeans
from sklearn.cluster import KMeans
import numpy as np
import matplotlib.pyplot as plt
mark=np.array([[35],[40],[50],[55],[80],[85],[90]])
model=KMeans(n_clusters=2, random_state=50)
model.fit(mark)
labels=model.labels_
centroids = model.cluster_centers_
print(centroids)
```

```python
plt.scatter(mark, labels)

plt.scatter(centroids, [10, 100], marker='+')

plt.xlabel("mark")
plt.ylabel("cl")
plt.title("kmean")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage

user = np.array([[10], [25], [38], [50]])

hie = linkage(user, method='single')
print(hie)
```

```python
plt.figure()
dendrogram(hie, labels=[10,25,38,50])
plt.title("Hierarchical Clustering Dendrogram")
plt.xlabel("user")
plt.ylabel("limits")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import linkage, dendrogram
user = np.array([[10], [20], [40], [60]])

hie = linkage(user, method='single')
print("Linkage Matrix:")
print(hie)

plt.figure()
dendrogram(hie, labels=[10,20,40,60])
plt.title("Divisive Hierarchical Clustering ")
plt.xlabel("mark")
plt.ylabel("limits")
plt.show()
```
