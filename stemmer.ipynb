```python
import nltk
nltk.download('punkt')
nltk.download('wordnet')
import nltk
nltk.download('punkt_tab')
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
give_a_command= input("live your command here:")
tokens = word_tokenize(give_a_command.lower())
lemmatizer = word_tokenize()
lemmas = [lemmatizer.lemmatizer(word) for word in tokens]
print("Lemmas:", lemmas)
positive = ["good", "love", "happy", "like", "nice"]
negative = ["bad", "wast of cost", "worst", "notgood", "poor"]
pos=0
neg=0
for word in lemmas:
    if word in positive:
        pos += 1
    elif word in negative:
        neg += 1
if pos > neg:
    print("reiv: POSITIVE")
elif neg > pos:
    print("reiv: NEGATIVE ")
else:
    print("reiv: NEUTRAL ")
```
