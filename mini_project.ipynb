```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
d=pd.read_csv("BIKE_DETAILS[1].csv")
print(d)
```

```python
df=d.dropna()
df
```

```python
df.drop_duplicates()
df
```

```python
le = LabelEncoder()
df["owner"] = le.fit_transform(df["owner"])
```

```python
X = df[["year", "km_driven", "owner"]]
y = df["selling_price"]
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X, y)

# 
sample = [[2019, 350, 1]]   # 0 = encoded va
prediction = model.predict(sample)

print("Predicted Price of the bike:", prediction[0])
```
